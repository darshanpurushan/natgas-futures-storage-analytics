# Natural Gas Futures Curve & Storage Analytics
Run the cells below to reproduce the demo research workflow.

```python
import sys
sys.path.append('../src')
from natgas_analytics.data import make_demo_curve
from natgas_analytics.storage import make_demo_storage, storage_features
from natgas_analytics.curve import add_curve_metrics
from natgas_analytics.backtest import run_backtest
```

```python
curve = make_demo_curve()
storage = make_demo_storage()
curve_metrics = add_curve_metrics(curve).reset_index(names='date')
storage_df = storage_features(storage)
curve_metrics.head()
```

```python
import pandas as pd
research = pd.merge_asof(curve_metrics.sort_values('date'), storage_df.sort_values('date'), on='date', direction='backward').dropna(subset=['storage_surprise_z','calendar_spread'])
results, metrics = run_backtest(research)
metrics
```

```python
results.set_index('date')['equity'].plot(figsize=(10,5), title='Strategy Equity')
```
